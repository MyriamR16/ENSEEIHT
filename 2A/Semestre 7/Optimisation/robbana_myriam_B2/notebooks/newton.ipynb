## Réponse à la Question 1

Interessons nous à la fonction $f_0(x) = sin(x)$.

La solution attendue pour cette fonction, peut importe le point initial, est : $solution = -pi/2$.
Ce résultat est cohérent car, en effet, $$ gradf_0(x) =  cos(x)$$ et $cos(-pi/2) = 0$ 

**Exemple 1 :** 

Pour cet exemple, on a $x_0 = -\pi/2$. On remarque que la solution attendue est identique à la valeur de départ. Le résultat obtenu est identique à cette valeur étant donné que $x_sol = -\pi/2$. 
Ce résultat est cohérent car, en effet, grad_f0(x) =  cos(x), et cos($-\pi/2$) ~= 0 < (tol_abs = 1e-10). Ainsi, on atteint la convergence avant même de pénêtrer dans la boucle **while**, d'où le flag qui vaut 0 et le nombre d'itération qui vaut 0 également.

**Exemple 2 :** 

Pour cet exemple, on a $x0 = -\pi/2 + 0.5$. Le résultat obtenu à la fin de l'algorithme  est identique à cette valeur étant donné que x_sol = $- \pi/2$. 
Ce résultat est cohérent car, en effet, la condition **if** précedant la boucle **while** n'est pas valide, étant donnée que x0 est différent de la solution attendue. Ainsi, on atteint la convergence après quelques passages dans la boucle **while**, et au bout de 3 passages correspondants à la valeur obtenue pour nb_iters, x_sol finit par converger vers la solution d'où le flag=0.

**Exemple 3 :** 

Pour cet exemple, on a $x_0 = \pi/2$ qui est l'opposé de la solution attendue. Le résultat obtenu à la fin de l'algorithme  est x_sol = $\pi/2$ qui est l'opposée de la solution souhaitée. 
Cela pourrait être dû à un point de symétrie où le point critique $\pi/2$ a une valeur symétrique équivalente. 
La CN1 manipulant uniquement des normes, est donc vérifée avec $norm(-\pi/2) = \pi/2$, d'où le flag=0.

## Réponse à la Question 2


Dans cette question, on s'interesse à la fonction $f_{1}$ :

$$
    f_{1}(x_1, x_2, x_3) = 2 (x_1 +x_2 + x_3 -3)^2 + (x_1-x_2)^2 + (x_2 - x_3)^2
$$

En calculant le gradiant de $f_{1}$ on obtient : 

grad_fct1(x) =  
$$\begin{pmatrix}
       4(x_1 + x_2 + x_3 - 3) + 2(x_1 - x_2) \\
       4(x_1 + x_2 + x_3 - 3) - 2(x_1 - x_2) +2(x_2 - x_3) \\
       4(x_1 + x_2 + x_3 - 3) - 2(x_2 - x_3)
\end{pmatrix}$$

On remarque que ce gradient s'annule si $x_1 = x_2 = x_3 = 1$.

En calculant la hessienne de $f_{1}$ on obtient : 

hess_fct1(x) =  
$$\begin{pmatrix}
       6 && 2 && 4 \\
       2 && 8 && 2 \\
       4 && 2 && 6
\end{pmatrix}$$

Les valeurs propres de cette matrice étant postives, la matrice hessienne est définie positive. De plus, cette matrice ne possède que des coefficients constants. 

Ainsi, la méthode de Newton converge en une seule itération pour une fonction quadratique ayant une Hessienne définie positive et constante. Ainsi, pour la fonction $f_{1}$, l'algorithme converge bien au bout d'une itération. 

## Réponse à la Question 3


Dans cette question on s'interesse à la fonction $f_{2}$ : 
$$
    f_{2}(x_1,x_2) = 100(x_2-x_1^2)^2 + (1-x_1)^2.
$$

En calculant le gradiant de $f_{2}$ on obtient la matrice ci-dessous : 

grad_fct2(x) =  
$$\begin{pmatrix}
       -400.x_1.(x_2 - x_1²) - 2(1 - x_1) \\
        200(x_2 - x_1²)
\end{pmatrix}$$

On peut remarquer que ce gradient s'annule lorsque $x_2 = x_1 = 1$.

En calculant la hessienne de $f_{2}$ on obtient : 

hess_fct2(x) =  
$$\begin{pmatrix}
       -400.(x_2 - 3.x_1²) + 2 && -400.x_1\\
       -400.x_1 && 200
\end{pmatrix}$$

On peut donc constater que les valeurs de la matrice hessienne de $f_{2}$ ne sont pas constantes et dépendent de x_1 et x_2.

Pour conclure, la matrice hessienne n'étant pas constante, l'approximation locale réalisée dans l'algorithme peut augmenter la fonction au lieu de la diminuer donc empêcher la convergence.
